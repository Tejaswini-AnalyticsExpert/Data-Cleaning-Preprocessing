1. Load Raw Data **bold text**

```
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Define the file paths (update folder name if needed)
file1 = 'Fitness Classes Data/Classes April-May 2018.csv'
file2 = 'Fitness Classes Data/Classes June 2018.csv'

# Read the collected data into pandas DataFrames
df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)

# Combine the two datasets into one
df = pd.concat([df1, df2], ignore_index=True)

print(f"Data loaded. Total rows: {len(df)}")
df.head()
```

2. Clean **Data**

```
# 1. Remove duplicates
initial_count = len(df)
df = df.drop_duplicates()
print(f"Removed {initial_count - len(df)} duplicate rows.")

# 2. Handle missing values
# Checking for nulls
print("Missing values before cleaning:\n", df.isnull().sum())

# We will fill missing Price values with the median price
# (or use df.dropna() if you're not allowed to guess values)
df['Price (INR)'] = df['Price (INR)'].fillna(df['Price (INR)'].median())

# 3. Correct data types
# Ensure Price is a float for calculations
df['Price (INR)'] = df['Price (INR)'].astype(float)

print("\nData cleaning step finished.")
```

3. Transform **Data**

```
# 1. Standardize date formats

date_col = 'BookingEndDateTime (Month / Day / Year)'
df[date_col] = pd.to_datetime(df[date_col], dayfirst=False)

# 2. Normalize or scale numerical features
# We will scale 'MaxBookees' and 'Number Booked' using StandardScaler
scaler = StandardScaler()
cols_to_scale = ['MaxBookees', 'Number Booked']

# Creating a new column for scaled data so we don't lose the original numbers
df[['MaxBookees_Scaled', 'Number Booked_Scaled']] = scaler.fit_transform(df[cols_to_scale])

print("Data transformation complete!")
df.head()
```

**1. Descriptive Statistics**

```
# Get summary statistics for numerical columns
stats_summary = df.describe()
print("Descriptive Statistics:")
display(stats_summary)
```

**2. Data Visualization**

```
import matplotlib.pyplot as plt
import seaborn as sns

# Set the visual style
sns.set(style="whitegrid")

# Create a figure with two subplots
plt.figure(figsize=(14, 6))

# Histogram: Distribution of Number Booked
plt.subplot(1, 2, 1)
sns.histplot(df['Number Booked'], bins=20, kde=True, color='skyblue')
plt.title('Distribution of Class Bookings')
plt.xlabel('Number of People Booked')

# Boxplot: Price (INR) to identify outliers
plt.subplot(1, 2, 2)
sns.boxplot(x=df['Price (INR)'], color='lightcoral')
plt.title('Price Distribution and Outliers')
plt.xlabel('Price (INR)')

plt.tight_layout()
plt.show()
```

**3. Correlation Analysis**

```
# Select only numerical columns for correlation
numerical_df = df.select_dtypes(include=['float64', 'int64'])

# Calculate correlation matrix
correlation_matrix = numerical_df.corr()

# Visualize with a Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```


 **Outliers: In the Boxplot, any dots far to the right are "outliers" (classes that are much more expensive than the rest).**

**Trends: In the Histogram, check if most classes are full (near the MaxBookees value) or mostly empty.**

**Correlation: A number close to 1.0 means a strong positive relationship, while a number close to -1.0 means a strong negative relationship.**
